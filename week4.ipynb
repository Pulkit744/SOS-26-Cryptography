```python
import hashlib
import hmac
```

```python
SECRET_KEY = b"supersecret"
```

```python
def insecure_mac(message):
    return hashlib.sha1(
        SECRET_KEY + message
    ).hexdigest()
```

```python
message = b"user=alice"

original_mac = insecure_mac(message)

print("\nInsecure SHA1(key||message) MAC:")
print(original_mac)
```

```python
import hashpumpy
```

```python
def secure_hmac(message):
    return hmac.new(
        SECRET_KEY,
        message,
        hashlib.sha1
    ).hexdigest()
```

```python
original_hmac = secure_hmac(message)

print("\nOriginal HMAC:")
print(original_hmac)
```

```python
extra_data = b"&admin=true"

forged_message = message + extra_data

print("\nForged Message:")
print(forged_message)
```

```python
actual_hmac = secure_hmac(forged_message)

print("\nOriginal HMAC:")
print(original_hmac)

print("\nHMAC of Modified Message:")
print(actual_hmac)
```

```python
if original_hmac == actual_hmac:
    print("\nAttack Successful")
else:
    print("\nAttack Failed")
```
