```python
import hashlib
import secrets
```

```python
key = secrets.token_bytes(16)
```

```python
def mac(message):
    return hashlib.sha1(key + message).hexdigest()
```

```python
message = b"Implementing"
```

```python
tag = mac(message)
```

```python
print("MAC:", tag)
print(mac(message) == tag)
print(mac(b"hacked") == tag)
```
