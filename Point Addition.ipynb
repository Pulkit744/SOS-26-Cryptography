```python
def point_addition(P, Q, a, p):
    if P is None:
        return Q
    if Q is None:
        return P
        
    x1, y1 = P
    x2, y2 = Q
    
    if x1 == x2 and (y1 + y2) % p == 0:
        return None
        
    if P != Q:
        num = (y2 - y1) % p
        den = (x2 - x1) % p
        lam = (num * pow(den, -1, p)) % p
    else:
        num = (3 * pow(x1, 2, p) + a) % p
        den = (2 * y1) % p
        lam = (num * pow(den, -1, p)) % p
        
    x3 = (pow(lam, 2, p) - x1 - x2) % p
    y3 = (lam * (x1 - x3) - y1) % p
    
    return (x3, y3)

def solve():
    p = 9739
    a = 497
    
    P = (493, 5564)
    Q = (1539, 4742)
    R = (4403, 5202)
    
    P2 = point_addition(P, P, a, p)
    P2_Q = point_addition(P2, Q, a, p)
    S = point_addition(P2_Q, R, a, p)
    
    print(S)

if __name__ == "__main__":
    solve()
```
