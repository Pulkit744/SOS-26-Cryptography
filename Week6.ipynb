```python
p = 97
a = 2
b = 3
G = (3, 6)
O = None
```

```python
def inv_mod(x, p):
    return pow(x, -1, p)

def point_add(P, Q):
    if P is None:
        return Q
    if Q is None:
        return P

    x1, y1 = P
    x2, y2 = Q

    if x1 == x2 and (y1 + y2) % p == 0:
        return None

    if P != Q:
        m = ((y2 - y1) * inv_mod(x2 - x1, p)) % p
    else:
        m = ((3 * x1 * x1 + a) * inv_mod(2 * y1, p)) % p

    x3 = (m * m - x1 - x2) % p
    y3 = (m * (x1 - x3) - y1) % p

    return (x3, y3)
```

```python
def scalar_mult(k, P):
    R = None
    Q = P

    while k:
        if k & 1:
            R = point_add(R, Q)
        Q = point_add(Q, Q)
        k >>= 1

    return R
```

```python
alice_private = 15
bob_private = 22

alice_public = scalar_mult(alice_private, G)
bob_public = scalar_mult(bob_private, G)

alice_shared = scalar_mult(alice_private, bob_public)
bob_shared = scalar_mult(bob_private, alice_public)

print("Alice Public:", alice_public)
print("Bob Public:", bob_public)
print("Alice Shared:", alice_shared)
print("Bob Shared:", bob_shared)
```

```python
def brute_force_dlp(P, G, limit=200):
    R = None

    for k in range(1, limit + 1):
        R = point_add(R, G)
        if R == P:
            return k

    return None
```

```python
recovered = brute_force_dlp(alice_public, G)

print("Recovered Private Key:", recovered)
print("Actual Private Key:", alice_private)
```
