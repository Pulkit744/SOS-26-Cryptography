```python
def caesar_encrypt(text, shift):
    result = ""

    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            new_char = chr((ord(char) - base + shift) % 26 + base)
            result += new_char
        else:
            result += char

    return result
```

```python
def caesar_decrypt(text, shift):
    return caesar_encrypt(text, -shift)
```

```python
message = "ATTACK DEFEND"
encrypted_caesar = caesar_encrypt(message, 3)
print(encrypted_caesar)

decrypted_caesar = caesar_decrypt(encrypted_caesar, 3)
print(decrypted_caesar)
```

```python
def vigenere_encrypt(text, key):
    result = ""
    key = key.upper()
    key_index = 0

    for char in text:
        if char.isalpha():
            shift = ord(key[key_index % len(key)]) - ord('A')

            base = ord('A') if char.isupper() else ord('a')
            new_char = chr((ord(char) - base + shift) % 26 + base)

            result += new_char
            key_index += 1
        else:
            result += char

    return result
```

```python
def vigenere_decrypt(text, key):
    result = ""
    key = key.upper()
    key_index = 0

    for char in text:
        if char.isalpha():
            shift = ord(key[key_index % len(key)]) - ord('A')

            base = ord('A') if char.isupper() else ord('a')
            new_char = chr((ord(char) - base - shift) % 26 + base)

            result += new_char
            key_index += 1
        else:
            result += char

    return result
```

```python
message = "ATTACKATDAWN"
key = "HELLO"

encrypted = vigenere_encrypt(message, key)
print(encrypted)

decrypted = vigenere_decrypt(encrypted, key)
print(decrypted)
```

```python
from collections import Counter
import string

def frequency_analysis(text):
    text = ''.join(c.upper() for c in text if c.isalpha())

    total = len(text)
    counts = Counter(text)

    print("\nLetter Frequencies:")
    print("-------------------")

    for letter in string.ascii_uppercase:
        freq = counts[letter]
        percent = (freq / total) * 100 if total > 0 else 0

        print(f"{letter}: {freq:3d} ({percent:5.2f}%)")

    print("\nMost Common Letters:")
    for letter, count in counts.most_common(5):
        print(f"{letter}: {count}")
```

```python
def guess_caesar_shift(ciphertext):
    ciphertext = ''.join(c.upper() for c in ciphertext if c.isalpha())
    counts = Counter(ciphertext)

    most_common = counts.most_common(1)[0][0]

    shift = (ord(most_common) - ord('E')) % 26

    print("Most common ciphertext letter:", most_common)
    print("Guessed shift:", shift)

    return shift
```

```python
def index_of_coincidence(text):
    text = ''.join(c.upper() for c in text if c.isalpha())
    N = len(text)

    counts = Counter(text)

    numerator = sum(f * (f - 1) for f in counts.values())
    denominator = N * (N - 1)

    return numerator / denominator if denominator else 0
```

```python
message = "THIS IS A SECRET MESSAGE"

caesar_cipher = caesar_encrypt(message, 7)
vig_cipher = vigenere_encrypt(message, "vigenere")
```

```python
print("Caesar IoC:",
      index_of_coincidence(caesar_cipher))

print("Vigenère IoC:",
      index_of_coincidence(vig_cipher))
```
