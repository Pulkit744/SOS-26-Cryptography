```python
import requests
import json

BASE_URL = "https://aes.cryptohack.org/block_cipher_starter"

def get_encrypted_flag():
    response = requests.get(f"{BASE_URL}/encrypt_flag/")
    return response.json()["ciphertext"]

def decrypt_ciphertext(ciphertext):
    response = requests.get(f"{BASE_URL}/decrypt/{ciphertext}/")
    return response.json()["plaintext"]

def solve():
    encrypted_flag = get_encrypted_flag()
    print(f"Encrypted flag (hex): {encrypted_flag}")

    decrypted_hex = decrypt_ciphertext(encrypted_flag)
    print(f"Decrypted (hex): {decrypted_hex}")

    flag = bytes.fromhex(decrypted_hex).decode('utf-8')
    print(f"Flag: {flag}")

if __name__ == "__main__":
    solve()
```
