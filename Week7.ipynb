```python
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
```

```python
key = get_random_bytes(16)
nonce = get_random_bytes(12)

plaintext = b"Hello AES-GCM!"
aad = b"authenticated data"

cipher = AES.new(key, AES.MODE_GCM, nonce=nonce)
cipher.update(aad)

ciphertext, tag = cipher.encrypt_and_digest(plaintext)

print("Ciphertext:", ciphertext.hex())
print("Tag:", tag.hex())
```

```python
cipher = AES.new(key, AES.MODE_GCM, nonce=nonce)
cipher.update(aad)

decrypted = cipher.decrypt(ciphertext)

cipher_verify = AES.new(key, AES.MODE_GCM, nonce=nonce)
cipher_verify.update(aad)

_, computed_tag = cipher_verify.encrypt_and_digest(decrypted)

print("Computed Tag:", computed_tag.hex())
print("Received Tag:", tag.hex())

if computed_tag == tag:
    print("Authentication Successful")
    print("Plaintext:", decrypted.decode())
else:
    print("Authentication Failed")
```
